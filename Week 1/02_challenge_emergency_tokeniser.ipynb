## The problem

In the lab, our basic whitespace script treats **`help!!!`** as **one single token**.

Why? The lab's cleaning line removes only `.` and `?`. The exclamation marks are neither removed nor separated, so `split()` has no space to cut at, and `help!!!` stays whole.

That matters for two reasons:

1. `help!!!` is a **different token** from `help`. A model that has seen `help` many times has no link to `help!!!`.
2. When the punctuation is glued to the word, the model may not recognise how critical the situation is. A panicked patient's urgency signal is hidden.

## Your task

Update the cleaning logic so that **exclamation marks are treated as separate tokens**:

```
'help!!!'  ->  ['help', '!', '!', '!']
```

This guarantees our system flags the urgent distress signals.
