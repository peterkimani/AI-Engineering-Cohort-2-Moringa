## The problem

In the lab, our basic whitespace script treats **`help!!!`** as **one single token**.

Why? The lab's cleaning line removes only `.` and `?`. The exclamation marks are neither removed nor separated, so `split()` has no space to cut at, and `help!!!` stays whole.

That matters for two reasons:

1. `help!!!` is a **different token** from `help`. A model that has seen `help` many times has no link to `help!!!`.
2. When the punctuation is glued to the word, the model may not recognise how critical the situation is. A panicked patient's urgency signal is hidden.

## Your task

Update the cleaning logic so that **exclamation marks are treated as separate tokens**:

```
'help!!!'  ->  ['help', '!', '!', '!']
```

This guarantees our system flags the urgent distress signals.

## 3. The solution

Below is the complete function. It keeps the starter's structure and fills in the three TODOs. The docstring says what the function does and how each step works.

Run the cell. The test loop at the bottom is the same as the starter's.

````python
import re
import numpy as np

def emergency_tokeniser(text):
    r"""
    What it does:
        Splits a message into tokens, keeping every '!' as its own token and
        removing all other punctuation.

    How it works:
        1. text.lower() makes every letter lowercase.
        2. text.replace("!", " ! ") puts a space on both sides of each '!', so
           'help!!!' becomes 'help ! ! ! ' and split() can later see each '!' alone.
        3. re.sub(r"[^\w\s!]", "", text) deletes every character that is NOT a word
           character (letter, digit or underscore), NOT whitespace, and NOT '!'.
        4. text.split() cuts on any run of whitespace and drops empty pieces.
        5. np.array(...) stores the tokens in a NumPy array.

    Parameters:
        text (str): the raw patient message.

    Returns:
        np.ndarray: the tokens in order, with each '!' as a separate element.
    """
    text = text.lower()

    # TODO 1: Pad '!' with spaces so split() treats them as separate tokens.
    text = text.replace("!", " ! ")

    # TODO 2: Remove any other punctuation that should not be a token.
    text = re.sub(r"[^\w\s!]", "", text)

    # TODO 3: Split on whitespace and return a NumPy array.
    tokens = np.array(text.split())
    return tokens

# Test cases
samples = ["Help!!!", "I cannot breathe!", "My chest hurts."]
for s in samples:
    print(s, "->", emergency_tokeniser(s))

# Expected for "Help!!!": ['help', '!', '!', '!']
````

### Expected output

```

I cannot breathe! -> ['i' 'cannot' 'breathe' '!']
My chest hurts. -> ['my' 'chest' 'hurts']
```

NumPy prints arrays without commas. That is only a display difference from the lesson's expected `['help', '!', '!', '!']`. The values are the same.

### Line-by-line explanation of the solution

**`import re`**
Loads Python's built-in regular-expression module. We need it for `re.sub`.

**`text = text.lower()`**
Unchanged from the starter. Everything becomes lowercase.

**TODO 1: `text = text.replace("!", " ! ")`**
`replace` swaps every `!` for `" ! "`: an exclamation mark with a space on each side. The lab used `replace` to **delete** characters (swap them for nothing). Here it **adds spaces** instead. This is the key idea of the challenge. After this line:

- `'help!!!'` becomes `'help ! ! ! '`
- `'i cannot breathe!'` becomes `'i cannot breathe ! '`

Extra spaces do no harm, because `split()` ignores repeated whitespace.

**TODO 2: `text = re.sub(r"[^\w\s!]", "", text)`**
`re.sub(pattern, replacement, text)` finds everything that matches the pattern and swaps it for the replacement. Here the replacement is `""`, so matches are **deleted**. Reading the pattern `[^\w\s!]` piece by piece:

| Part | Meaning |
|---|---|
| `[ ... ]` | match any one character from this set |
| `^` as the first character inside the brackets | **not**: invert the set, so match anything *except* the listed characters |
| `\w` | a word character: a letter, digit or underscore |
| `\s` | whitespace (space, tab, newline) |
| `!` | a literal exclamation mark |

So the pattern means: *"any character that is not a letter or digit, not whitespace, and not an exclamation mark."* All of those get deleted, which removes `.`, `?`, `,`, `;`, brackets and so on, but keeps the `!` we just padded. The `r` in front of the string makes it a **raw string**, so Python leaves the backslashes alone for the regex engine to read.

**TODO 3: `tokens = np.array(text.split())`**
`split()` with no argument cuts at any whitespace and returns a list of tokens. `np.array(...)` turns the list into a NumPy array, as in the lab. Each `!` is now its own element because it was surrounded by spaces.

**`return tokens`**
Sends the array back to the caller.

### Trace: what happens to `"Help!!!"` at each step

| Step | Value of `text` |
|---|---|
| start | `'Help!!!'` |
| after `lower()` | `'help!!!'` |
| after `replace("!", " ! ")` | `'help ! ! ! '` |
| after `re.sub(...)` (nothing to remove) | `'help ! ! ! '` |
| after `split()` | `['help', '!', '!', '!']` |
| after `np.array(...)` | `array(['help', '!', '!', '!'])` |

And for `"My chest hurts."`: the full stop is not a letter, digit, space or `!`, so `re.sub` deletes it, leaving `'my chest hurts'`, which splits into `['my', 'chest', 'hurts']`.

### Why the order matters

- **Pad before you split**, because `split()` can only cut at spaces that already exist.
- **Remove other punctuation before you split**, so stray marks never become tokens.
- **The `!` survives the cleaning step** only because the pattern explicitly allows it. If you used a pattern that removed every non-letter, the `!` would be deleted and there would be nothing left to flag.

### Alternative: chained `replace` calls

The lab removed `.` and `?` one at a time with `replace`. You could extend that approach: pad the `!`, then call `replace` once per mark you want to remove. It works and is easy to read for a short list. The regular expression is shorter and also handles marks you did not think of, which is why the solution uses it. Both approaches are acceptable.

### A limitation to discuss

The pattern deletes the apostrophe too, so `can't` becomes `cant` and `it's` becomes `its`. That is fine for this exercise but is a real trade-off in production. Real tokenisers such as `tiktoken` handle contractions and punctuation with learned rules instead of a hand-written pattern.
