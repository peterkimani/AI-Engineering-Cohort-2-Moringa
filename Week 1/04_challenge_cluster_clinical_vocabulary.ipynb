## 3. The solution

Below is the complete code. It keeps the starter's structure and fills in the three TODOs. The `cosine_similarity` function has a docstring that says what it does and how the calculation works. Run the cell.

````python
import numpy as np

embeddings = {
    "fever":     np.array([0.9, 0.0, 0.3]),
    "pain":      np.array([0.8, 0.1, 0.4]),
    "discharge": np.array([0.5, 0.5, 0.1]),
    "billing":   np.array([0.0, 0.9, 0.1]),
    "admit":     np.array([0.2, 0.8, 0.2]),
    # TODO 1: Add a 'cough' vector that sits close to 'fever' and far from 'billing'.
    "cough":     np.array([0.85, 0.0, 0.35]),
}

def cosine_similarity(word1, word2):
    """
    What it does:
        Measures how close two words are in embedding space.

    How the calculation works:
        1. Look up both vectors in the embeddings dictionary.
        2. np.dot multiplies the vectors position by position and adds the products.
        3. np.linalg.norm gives each vector's length (square root of the sum of squares).
        4. Dividing the dot product by (length1 * length2) cancels out vector size,
           leaving only the angle between the vectors. 1.0 means the same direction,
           0 means unrelated.
    """
    vec1, vec2 = embeddings[word1], embeddings[word2]
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

# TODO 2: Print cosine_similarity('cough', 'fever') and cosine_similarity('cough', 'billing').
fever_score = cosine_similarity('cough', 'fever')
billing_score = cosine_similarity('cough', 'billing')
print(f"cough vs fever  : {fever_score:.4f}")
print(f"cough vs billing: {billing_score:.4f}")

# TODO 3: Confirm the fever score is meaningfully higher than the billing score.
print("Fever score is higher than billing score:", fever_score > billing_score)
print(f"Gap between the two scores: {fever_score - billing_score:.4f}")
````


### Line-by-line explanation

**TODO 1: `"cough": np.array([0.85, 0.0, 0.35]),`**
We added one entry to the dictionary. The word is the key, and the three numbers follow the axes **[Clinical, Administrative, Urgency]**:

- **0.85 clinical**: a cough is a clearly medical symptom, almost as clinical as `fever` (0.9) and `pain` (0.8).
- **0.0 administrative**: a cough has nothing to do with billing or hospital process.
- **0.35 urgency**: between `fever` (0.3) and `pain` (0.4), reflecting "somewhat time-sensitive".

The comma at the end keeps the dictionary valid.

**TODO 2: the two scores**
`cosine_similarity('cough', 'fever')` looks up both vectors and returns their cosine. The result is stored in `fever_score`. The same is done for `billing` and stored in `billing_score`. The f-string `{fever_score:.4f}` prints each number with 4 digits after the decimal point.

**TODO 3: the comparison**
`fever_score > billing_score` is `True` when the first number is bigger. The last line also prints the **gap** between the scores, because "bigger" is not enough: we need it to be **meaningfully** bigger. A gap of about 0.96 is huge. A gap of 0.01 would mean the word sits almost halfway between the two clusters.

### The calculation worked by hand

**`cough` against `fever`** (`cough` = [0.85, 0.0, 0.35], `fever` = [0.9, 0.0, 0.3]):

1. Dot product: 0.85 x 0.9 + 0.0 x 0.0 + 0.35 x 0.3 = 0.765 + 0 + 0.105 = **0.87**
2. Length of `cough`: the square root of (0.7225 + 0 + 0.1225) = the square root of 0.845 = **0.9192**
3. Length of `fever`: the square root of (0.81 + 0 + 0.09) = the square root of 0.9 = **0.9487**
4. Cosine: 0.87 / (0.9192 x 0.9487) = 0.87 / 0.8720 = **0.9976**

**`cough` against `billing`** (`billing` = [0.0, 0.9, 0.1]):

1. Dot product: 0.85 x 0.0 + 0.0 x 0.9 + 0.35 x 0.1 = 0 + 0 + 0.035 = **0.035**
2. Length of `billing`: the square root of (0 + 0.81 + 0.01) = the square root of 0.82 = **0.9055**
3. Cosine: 0.035 / (0.9192 x 0.9055) = 0.035 / 0.8323 = **0.0420**

**Why the scores come out this way:** `cough` and `fever` have big numbers in the same positions (clinical) and near-zero numbers in the same position (administrative), so they point almost exactly the same way. `cough` and `billing` overlap only in the tiny urgency component, so they are almost at right angles.

### What the result tells us

`cough` sits firmly in the **clinical cluster**. That is what we want from an embedding: a guardrail that watches the region around `fever` and `pain` will also catch `cough`.

### Other vectors that would also work

The answer is not unique. Any vector with a high first number, a near-zero second number and a small third number works, for example `[0.9, 0.05, 0.2]` or `[0.7, 0.0, 0.2]`. Cosine similarity compares direction only, so even `[8.5, 0.0, 3.5]` (ten times longer) would give the same scores as `[0.85, 0.0, 0.35]`.

Try replacing the `cough` vector in the solution cell with `[0.0, 0.9, 0.1]` and re-running it. That is the pattern of `billing`, so you should see `cough vs billing` jump to **1.0000** and `cough vs fever` fall to about **0.03**. The check in TODO 3 would then print `False`. This shows how the numbers, not the word, decide which cluster a word belongs to. Restore the original vector afterwards.
