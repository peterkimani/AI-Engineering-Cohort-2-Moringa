### How the starter code works

**`embeddings = { ... }`**
The same dictionary as in Lab 2: each word is a key, and its value is a 3-number NumPy array along the axes **[Clinical-Symptom, Administrative, Time/Urgency]**. The only difference is that the `discharge` comment is gone and there is a `TODO 1` comment where `cough` belongs. Note the trailing comma after the last entry: it is fine in Python, and it means you can add a new line below without editing the line above.

**`def cosine_similarity(word1, word2):`**
The same function as in Lab 2, written more compactly:

- `vec1, vec2 = embeddings[word1], embeddings[word2]` looks up both vectors in one line.
- `np.dot(vec1, vec2)` multiplies the vectors position by position and adds the products.
- `np.linalg.norm(vec)` gives a vector's length.
- Dividing the dot product by the two lengths leaves only the angle between the vectors. A score near 1.0 means "very similar meaning", and a score near 0 means "unrelated".

## How to choose a vector for `cough` (TODO 1)

Choosing a vector is a design decision. Look at the existing words:

| Word | Clinical | Administrative | Urgency |
|---|---|---|---|
| fever | 0.9 | 0.0 | 0.3 |
| pain | 0.8 | 0.1 | 0.4 |
| billing | 0.0 | 0.9 | 0.1 |
| admit | 0.2 | 0.8 | 0.2 |

Ask three questions about `cough`:

1. **Is it a medical symptom?** Yes, so its clinical number should be **high**.
2. **Is it about paperwork, money or hospital process?** No, so its administrative number should be **low**.
3. **Is it time-critical?** Somewhat. A cough is less urgent than chest pain but not zero, so urgency should be **low to medium**.

So `cough` should have a **pattern similar to `fever`**: high first number, near-zero second number, small third number. Because cosine similarity compares *direction*, the pattern matters more than the exact size of the numbers.
